# Baseline exploration

This notebook records the completed Stage 1 baseline for the Statlog German Credit dataset (UCI ID 144). It reads the saved metrics and does not fetch data or retrain the model, so opening it cannot replace the preserved baseline artifacts.

Target mapping: `1 = Good credit risk`; `2 = Bad credit risk`. Financial error cost: `(false approvals × 5) + (false rejections × 1)`. A false approval is a bad-risk applicant predicted good; a false rejection is a good-risk applicant predicted bad.

See [`../docs/stage1_report.md`](../docs/stage1_report.md) for the original data exploration, plots, preprocessing, and full metric definitions.

```python
import json
from pathlib import Path
import pandas as pd

project_root = Path.cwd()
if not (project_root / 'models' / 'stage1_metrics.json').exists():
    project_root = project_root.parent
metrics_path = project_root / 'models' / 'stage1_metrics.json'
with metrics_path.open(encoding='utf-8') as f:
    saved = json.load(f)

rows = []
for model_name, result in saved['metrics'].items():
    rows.append({
        'Model': model_name.replace('_', ' ').title(),
        'Accuracy': result['accuracy'],
        'Balanced accuracy': result['balanced_accuracy'],
        'ROC-AUC': result.get('roc_auc'),
        'False approvals': result['false_approvals'],
        'False rejections': result['false_rejections'],
        'Financial error cost': result['error_cost'],
    })
baseline = pd.DataFrame(rows).set_index('Model')
baseline
```

## Preserved result

Logistic Regression achieved 78% accuracy, 70.95% balanced accuracy, 0.804 ROC-AUC, 28 false approvals, 16 false rejections, and weighted error cost 156. The all-good dummy achieved 70% accuracy, 60 false approvals, and cost 300. These are the original held-out results in `models/stage1_metrics.json`.

The baseline report describes aggregate predictive performance. No subgroup fairness audit or mitigation was part of Stage 1.
